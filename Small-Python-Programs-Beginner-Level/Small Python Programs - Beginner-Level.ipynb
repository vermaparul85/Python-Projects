# 3. Generate Fibonacci Sequence

## Method 1

```python
def fabonacci(n):
    fab = [1]
    for number in range(n):
        if len(fab) == 1:
            fab.append(1)
        else:
            fab.append(fab[-2] + fab[-1])
        
    return fab
```

```python
fabonacci(15)
```

## Method 2 - Using recursive method

```python
def fabonacci(n):
    if n <= 1:
        return 1
    else:
        return fabonacci(n-1) + fabonacci(n-2)
```

```python
fabonacci(10)
```

* This program find prime numbers until the user chooses to stop asking for the next one.

```python
def is_prime(num):
    prime_flag = True
    for i in range(2,num):
        if num % i == 0:
            prime_flag = False
    return prime_flag

def next_prime_number(num):
    while True:
        if is_prime(num):
            return num
        else:
            num += 1

next_num = 1
while True:
    user_input = input('Do you want next prime number? Type Y or N ').lower()
    if user_input == 'y':
        next_num = next_prime_number(next_num + 1)
        print(next_num)
    else:
        break
```

# 7. Change Return Program
* The user enters a cost of an item and amount of money given. 
* The program will figure out the change and the number of quarters, dimes, nickels, pennies needed for the change.

```python
def calculate_change(cost,payment):
    quarters = 0
    dimes = 0
    nickels = 0
    pennies = 0
    
    change = round(payment - cost,2)
    print(f'Change : {change}')
    
    while round(change,2) >= quarters_value:
        quarters += 1
        change -= quarters_value

    while round(change,2) >= dimes_value:
        dimes += 1
        change -= dimes_value

    while round(change,2) >= nickels_value:
        nickels += 1
        change -= nickels_value

    while round(change,2) >= pennies_value:
        pennies += 1
        change -= pennies_value

    print(f'Quarters : {quarters}')
    print(f'Dimes : {dimes}')
    print(f'Nickels : {nickels}')
    print(f'Pennies : {pennies}')
    
quarters_value = .25
dimes_value = .1
nickels_value = .05
pennies_value = .01

calculate_change(3.59,5)
```

# 10. Find Factorial of a Given Number

# 14. Check whether the number is armstrong number or not
Armstorng number : if the number is equal to the sum of its own digits, each raised to power equal to the number of digits in the number.

```python
def armstrong_number(num):
    num_str = str(num)
    num_digits = len(num_str)
    total_sum = 0
    for i in num_str:
        total_sum += int(i) ** num_digits
    return num == total_sum
```

```python
num = int(input('Enter a number: '))
if armstrong_number(num):
    print(f'{num} is a armstrong number.')
else:
    print(f'{num} is not a armstrong number.')
```

## Find the armstrong numbers for a given range

```python
num_list = [num for num in range(1,10000) if armstrong_number(num)]
print(num_list)
```

# 15. Find LCM

```python
def compute_lcm(x, y):
    if x > y:
        greater = x
    else:
        greater = y

    while True:
        if greater % x == 0 and greater % y == 0:
            lcm = greater
            break
        else:
            greater += 1

    return lcm
```

```python
num1 = int(input('Enter the number: '))
num2 = int(input('Enter the number: '))
compute_lcm(num1, num2)
```

# 16. Find HCF

```python
def compute_hcf(x, y):
    if x > y:
        smaller = y
    else:
        smaller = x

    for i in range(1, smaller+1):
        if x % i == 0 and y % i == 0:
            hcf = i
    
    return hcf
```

```python
compute_hcf(54,24)
```

# 17. Convert Decimal to Binary, Octal, Hexadecimal

## Method 1 - using builtin functions

```python
num = int(input('Enter a number'))
print(f'Binary conversion: {bin(num)}')
print(f'Octal conversion: {oct(num)}')
print(f'Hex conversion: {hex(num)}')
```

## Method 2 - Creating user defined functions

```python
def decimal_to_binary(n):
    bin_str = ''
    while n > 0:
        bin_num = n % 2
        bin_str = str(bin_num) + bin_str
        n = n // 2
    return bin_str

def decimal_to_oct(n):
    oct_str = ''
    while n > 0:
        oct_num = n % 8
        oct_str = str(oct_num) + oct_str
        n = n // 8
    return oct_str

def decimal_to_hex(n):
    hex_form = {10:'a', 11:'b', 12:'c', 13:'d', 14:'e', 15:'f'}
    hex_str = ''
    while n > 0:
        hex_num = n % 16
        if hex_num in hex_form:
            hex_str = hex_form[hex_num] + hex_str
        else:
            hex_str = str(hex_num) + hex_str
        n = n // 16
    return hex_str
```

```python
num = int(input('Enter a number'))
print(f'Binary conversion: {decimal_to_binary(num)}')
print(f'Octal conversion: {decimal_to_oct(num)}')
print(f'Hex conversion: {decimal_to_hex(num)}')
```

# 21. Find whether the given number is disarium number or not.
A disarium number is a number that is equal to the sum of its digits each raised to the power of their respective positions in the number.

## Method 1

```python
def is_disarium(num):
    total_sum = 0
    for i, n in enumerate(str(num)):
        total_sum = total_sum + (int(n) ** (i+1))
    return num == total_sum
```

```python
num = int(input('Enter number: '))
if is_disarium(num):
    print(f'{num} is a disarium number.')
else:
    print(f'{num} is not a disarium number.')
```

## Method 2

```python
def is_disarium(num):
    total_sum = sum(int(n) ** (i+1) for i, n in enumerate(str(num)))
    return num == total_sum
```

```python
num = int(input('Enter number: '))
if is_disarium(num):
    print(f'{num} is a disarium number.')
else:
    print(f'{num} is not a disarium number.')
```

## Find the disarium numbers for a given range

```python
disarium_list = [num for num in range(1,300) if is_disarium(num)]
print(disarium_list)
```

# 27. Count Occurrences of an Element in a List

## Method 1 - using collections library

```python
from collections import Counter

def count_occurrences(l, element):
    counter = Counter(l)
    return counter[element]
```

```python
my_list = [1,2,3,4,2,5,2,3,4,6,5,5,5,5]
element_to_count = 5
result = count_occurrences(my_list, element_to_count)
print(f'The element {element_to_count} appears {result} times in the list')
```

## Method 2 - using list count method

```python
def count_occurrences(l, element):
    counter = l.count(element)
    return counter
```

```python
my_list = [1,2,3,4,2,5,2,3,4,6,5,5,5,5]
element_to_count = 5
result = count_occurrences(my_list, element_to_count)
print(f'The element {element_to_count} appears {result} times in the list')
```

# 36. Move Given Element at the End of the List

```python
def move_to_end(my_list, element):
    cnt = my_list.count(element)
    new_list = [i for i in my_list if i != element]
    new_list.extend([element] * cnt)
    return new_list

move_to_end([1,4,3,1,6,5], 1)
```

# 38. Multiply All Numbers in a List

## Method 1 - using for loop

```python
def multiply_numbers(num_list):
    result = 1
    for num in num_list:
        result *= num
    return result
```

```python
numbers = list(range(1,10))
result = multiply_numbers(numbers)
print(f'Input list: {numbers}')
print(f'Multiplication of all numbers in the list: {result}')
```

```python
numbers = [10, 2, 5, 25]
result = multiply_numbers(numbers)
print(f'Input list: {numbers}')
print(f'Multiplication of all numbers in the list: {result}')
```

## Method 2 - using functools library reduce function

```python
from functools import reduce

numbers = [10, 2, 5, 25]
result = reduce(lambda a, b: a * b, numbers)
print(f'Input list: {numbers}')
print(f'Multiplication of all numbers in the list: {result}')
```
